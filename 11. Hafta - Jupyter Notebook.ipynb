# Veri Setinin Yüklenmesi

```python
from skimage import data
image = data.camera()
```

# Edge Detection

## Sobel Operatörü

```python
from skimage import filters
import matplotlib.pyplot as plt

sobel_edges = filters.sobel(image)

plt.figure(figsize=(10,5))

plt.subplot(1,2,1)
plt.imshow(image, cmap="gray")
plt.title("Orijinal Görüntü")
plt.axis("off")

plt.subplot(1,2,2)
plt.imshow(sobel_edges, cmap="gray")
plt.title("Sobel Yöntemi")
plt.axis("off")

plt.show()
```

## Canny Operatörü

```python
from skimage import feature

canny_edges = feature.canny(image, sigma=2)

plt.figure(figsize=(10,5))

plt.subplot(1,2,1)
plt.imshow(image, cmap="gray")
plt.title("Orijinal")

plt.subplot(1,2,2)
plt.imshow(canny_edges, cmap="gray")
plt.title("Canny Yöntemi")
plt.axis("off")

plt.show()
```

# Morphological Operations

## Erosion Uygulaması

```python
from skimage import data, morphology
import matplotlib.pyplot as plt

image = data.binary_blobs(length=256)

selem = morphology.disk(3)

eroded = morphology.erosion(image, selem)

plt.figure(figsize=(10,5))

plt.subplot(1,2,1)
plt.imshow(image, cmap="gray")
plt.title("Orijinal")

plt.subplot(1,2,2)
plt.imshow(eroded, cmap="gray")
plt.title("Erosion")

plt.show()
```

## Dilation Uygulaması

```python
from skimage import data, morphology
import matplotlib.pyplot as plt

image = data.binary_blobs(length=256)

selem = morphology.disk(3)

dilated = morphology.dilation(image, selem)

plt.figure(figsize=(10,5))

plt.subplot(1,2,1)
plt.imshow(image, cmap="gray")
plt.title("Orijinal")

plt.subplot(1,2,2)
plt.imshow(dilated, cmap="gray")
plt.title("Dilation")

plt.show()
```

## Opening Uygulaması

```python
from skimage import data, morphology
import matplotlib.pyplot as plt

image = data.binary_blobs(length=256)

selem = morphology.disk(2)

opened = morphology.opening(image, selem)

plt.figure(figsize=(10,5))

plt.subplot(1,2,1)
plt.imshow(image, cmap="gray")
plt.title("Orijinal")

plt.subplot(1,2,2)
plt.imshow(opened, cmap="gray")
plt.title("Opening")

plt.show()
```
